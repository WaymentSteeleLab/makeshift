```python
import makeshift as ms
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv('data_hsqcs_24jun26_metadata.csv')
```

```python
pHs = []
for i, row in df.iterrows():
    entry = row['BMRB_entry']
    chemshift = row['chem_shift_id']
    entity = row['Entity_ID']
    cs = ms.ChemicalShifts.from_bmrb(entry)
    cond = cs.entry.sample_conditions()
    if 'pH' in cond.columns:
        pH = cond['pH'][0]
        pHs.append(pH)
    else:
        pHs.append('None')
```

```python
df['pH'] = pHs
df['pH'] = df['pH'].replace("None", float("nan"))
```

```python
df.to_csv('data_hsqcs_24jun26_metadata_v2.csv', index=False)
```
